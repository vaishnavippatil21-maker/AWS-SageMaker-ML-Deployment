```python
import boto3
import sagemaker
import pandas as pd
from sklearn.datasets import load_iris

print("AWS Region:", boto3.Session().region_name)
print("SageMaker Version:", sagemaker.__version__)

iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df["target"] = iris.target

print("Dataset Shape:", df.shape)
df.head()
```

```python
from sklearn.model_selection import train_test_split

session = sagemaker.Session()
role = sagemaker.get_execution_role()
bucket = session.default_bucket()

train, test = train_test_split(
    df, test_size=0.2, random_state=42
)

train.to_csv("train.csv", index=False)
test.to_csv("test.csv", index=False)

train_uri = session.upload_data(
    "train.csv", bucket=bucket, key_prefix="iris/train"
)

test_uri = session.upload_data(
    "test.csv", bucket=bucket, key_prefix="iris/test"
)

print("S3 Bucket:", bucket)
print("Training samples:", len(train))
print("Testing samples:", len(test))
print("Training data:", train_uri)
print("Testing data:", test_uri)
```

```python
%%writefile train.py

import os
import pandas as pd
import joblib

from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

train = pd.read_csv("/opt/ml/input/data/train/train.csv")
test = pd.read_csv("/opt/ml/input/data/test/test.csv")

X_train = train.drop("target", axis=1)
y_train = train["target"]

X_test = test.drop("target", axis=1)
y_test = test["target"]

model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

model.fit(X_train, y_train)

predictions = model.predict(X_test)
accuracy = accuracy_score(y_test, predictions)

print("Model Accuracy:", accuracy)

joblib.dump(model, "/opt/ml/model/model.joblib")


def model_fn(model_dir):
    return joblib.load(
        os.path.join(model_dir, "model.joblib")
    )
```

```python
from sagemaker.sklearn.estimator import SKLearn

estimator = SKLearn(
    entry_point="train.py",
    role=role,
    instance_count=1,
    instance_type="ml.m5.large",
    framework_version="1.2-1",
    py_version="py3",
    sagemaker_session=session,
    max_run=600,
    base_job_name="iris-random-forest"
)

print("Training configuration ready")
```

```python
estimator.fit({
    "train": train_uri,
    "test": test_uri
})
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report
import joblib

X_train = train.drop("target", axis=1)
y_train = train["target"]

X_test = test.drop("target", axis=1)
y_test = test["target"]

model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

model.fit(X_train, y_train)

predictions = model.predict(X_test)

print("Training completed successfully!")
print("Model Accuracy:", accuracy_score(y_test, predictions))
print(classification_report(y_test, predictions))

joblib.dump(model, "model.joblib")
print("Model saved successfully!")
```

```python
import xgboost as xgb
import tarfile
from sklearn.metrics import accuracy_score

xgb_model = xgb.XGBClassifier(
    n_estimators=50,
    max_depth=3,
    random_state=42,
    n_jobs=2,
    objective="multi:softmax",
    num_class=3
)

xgb_model.fit(X_train.to_numpy(), y_train)

predictions = xgb_model.predict(X_test.to_numpy())

print("Training successful!")
print("Accuracy:", accuracy_score(y_test, predictions))

xgb_model.get_booster().save_model("xgboost-model.json")

with tarfile.open("xgb-model.tar.gz", "w:gz") as tar:
    tar.add("xgboost-model.json", arcname="xgboost-model")

xgb_uri = session.upload_data(
    "xgb-model.tar.gz",
    bucket=bucket,
    key_prefix="iris/xgboost"
)

print("Model uploaded:", xgb_uri)
```

```python
image_uri = sagemaker.image_uris.retrieve(
    framework="xgboost",
    region="ap-south-1",
    version="3.0-5",
    image_scope="inference",
    instance_type="ml.m5.large"
)

print(image_uri)
print("Ready for deployment")
```

```python
from sagemaker.model import Model

deployment_model = Model(
    image_uri=image_uri,
    model_data=xgb_uri,
    role=role,
    sagemaker_session=session,
    name="iris-xgboost-model"
)

deployment_model.deploy(
    serverless_inference_config=serverless_config,
    endpoint_name="iris-xgboost-endpoint"
)

print("Deployment successful!")
```

```python
runtime = boto3.client("sagemaker-runtime")

response = runtime.invoke_endpoint(
    EndpointName="iris-xgboost-endpoint",
    ContentType="text/csv",
    Accept="text/csv",
    Body=b"5.1,3.5,1.4,0.2"
)

result = response["Body"].read().decode()

predicted_class = int(float(result.strip().strip("[]")))

print("Endpoint prediction:", result)
print("Predicted flower:", iris.target_names[predicted_class])
print("Cloud inference successful!")
```
